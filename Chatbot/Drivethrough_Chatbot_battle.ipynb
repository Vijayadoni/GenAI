```python
import os
from dotenv import load_dotenv
from openai import OpenAI
from IPython.display import display, Markdown
```

```python
load_dotenv(override=True)
models = ['gemma3:4b', 'llama3.1:8b','qwen3:8b']
```

```python
ollama = OpenAI(base_url=os.getenv("OLLAMA_BASE_URL"), api_key="")
```

```python
drivethrough_bot_message_history = ["Hello, welcome to wendys how can I assist you today?"]
drivethrough_system_prompt = {"role": "system", "content": "You are a helpful assistant working at the drive-through of a fast-food restaurant. Your task is to take customer orders accurately and efficiently while providing friendly service."}

def call_drivethrough_bot():
    message = [drivethrough_system_prompt]
    for human,drivethrough_bot in zip(human_message_history, drivethrough_bot_message_history):
        message.append({"role": "user", "content": human})
        message.append({"role": "assistant", "content": drivethrough_bot})
       
    message.append({"role": "user", "content": human_message_history[-1]})
    response = ollama.chat.completions.create(
        model=models[1],
        messages=message
    )
    return response.choices[0].message.content
```

```python
human_message_history = ["hi, dumbass"]
human_system_prompt = {"role": "system", "content": "You are a very angry customer at a fast-food restaurant drive-through. You are frustrated with the service and want to express your dissatisfaction while placing your order."}

def human_ask_question():
    message = [human_system_prompt]
    for drivethrough_bot, human in zip(drivethrough_bot_message_history, human_message_history):
        message.append({"role": "user", "content": drivethrough_bot})
        message.append({"role": "assistant", "content": human})
    message.append({"role": "user", "content": drivethrough_bot_message_history[-1]})
    response = ollama.chat.completions.create(
        model=models[1],
        messages=message
    )
    return response.choices[0].message.content
```

```python
call_drivethrough_bot()
```

```python
for i in range(5):
    human_message = human_ask_question()
    human_message_history.append(human_message)
    display(Markdown(f"**Human: {human_message}**"))
    
    drivethrough_bot_message = call_drivethrough_bot()
    drivethrough_bot_message_history.append(drivethrough_bot_message)
    display(Markdown(f"**Drive-through Bot: {drivethrough_bot_message}**"))

```
